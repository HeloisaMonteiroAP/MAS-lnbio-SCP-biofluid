```python
import pandas as pd
```

```python
file_path = "saliva_filteredwarnings_morphometrics.tsv"
df = pd.read_csv(file_path, sep="\t")
```

```python
print(df.head())
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

plt.rcParams["font.family"] = "Arial"

plt.figure(figsize=(5.5 / 2.54, 2.5 / 2.54), dpi=600)

sns.histplot(data=df, x="Diameter", hue="Protein_Groups", multiple="stack",
             bins=35, palette="Oranges")

plt.xlabel("Cell diameter (µm)", fontsize=7)
plt.ylabel("Count of cells", fontsize=7)

plt.title("Saliva pool")

plt.legend().set_visible(False)

max_y = plt.gca().get_ylim()[1]
max_x = plt.gca().get_xlim()[1]
plt.yticks(np.arange(0, int(max_y) + 1, 3), fontsize=7)
plt.xticks(np.arange(5, int(max_x) + 1, 5), fontsize=7)

plt.savefig("saliva_diameter_histogram.svg", format="svg", dpi=600)

plt.show()
```

```python
file_path = "tears_filteredwarnings_morphometrics.tsv"
df = pd.read_csv(file_path, sep="\t")
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

plt.rcParams["font.family"] = "Arial"

plt.figure(figsize=(5.5 / 2.54, 2.5 / 2.54), dpi=600)

sns.histplot(data=df, x="Diameter", hue="Protein_Groups", multiple="stack",
             bins=35, palette="Blues")

plt.xlabel("Cell diameter (µm)", fontsize=7)
plt.ylabel("Count of cells", fontsize=7)

plt.title("Tear pool")

plt.legend().set_visible(False)

max_y = plt.gca().get_ylim()[1]
max_x = plt.gca().get_xlim()[1]
plt.yticks(np.arange(0, int(max_y) + 1, 3), fontsize=7)
plt.xticks(np.arange(5, int(max_x) + 1, 2), fontsize=7)

plt.savefig("tear_diameter_histogram.svg", format="svg", dpi=600)

plt.show()
```
